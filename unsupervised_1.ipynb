```python
# position컬럼과 position_group컬럼의 값들이 범주형이고, 문자로 구성되어 lable encoding 후 상관관계 보기
from sklearn.preprocessing import LabelEncoder

le1 = LabelEncoder()
le2 = LabelEncoder()
position = le1.fit_transform(df['position'])
position_group = le2.fit_transform(df['position_group'])

corr = pd.DataFrame({
    'position': position,
    'position_group': position_group
}).corr()

print(corr)

```

```python
# 방향은 반대이나 매우 강한 상관관계를 보여주는 두 컬럼 중 한 컬럼 삭제
df.drop(['position_group'], axis=1, inplace=True)
df.info()
```

```python
# gender와 같은 자료를 0과 1로만 분류할 경우, 머신러닝 모델 입장에서는 숫자츼 크기가 의미를 가질 수 있다.
# 크기보다 범주자체에 집중하는 트리기반 모델은 비교적 안전하나 회귀모델과 거리기반 모델에서는 특정 성별이 더 크다는 의미없는 해석 가능성 존재
# one hot encoding을 통해 특정 성별이 우위에 있다거나, 크다는 해석이 없어져 안전
# 불필요한 중복정보가 들어가 약간 비효율적이며 다중공선성 발생하나, 직관적이고 해석도 쉽다는 측면에서 drop_first=False로 진행
df = pd.get_dummies(df, columns=['gender'], drop_first=False, dtype=int) # int미지정시 boolean자료로 나옴
df.head()
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
import pandas as pd
import umap
import numpy as np
import matplotlib.pyplot as plt

# 1. SBERT 모델 로딩
model = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')  # 다국어 지원

# 2. 개별 컬럼 임베딩
emb1 = model.encode(df['profile'], show_progress_bar=True)
emb2 = model.encode(df['episode'], show_progress_bar=True)

# 3. 두 임베딩을 이어붙이기 (concatenate)
combined_embeddings = np.concatenate([emb1, emb2], axis=1)

# 4. 클러스터링 (KMeans)
kmeans = KMeans(n_clusters=2, random_state=42)
df['cluster'] = kmeans.fit_predict(combined_embeddings)

# 5. UMAP 시각화
reducer = umap.UMAP(n_neighbors=5, random_state=42)
embedding_2d = reducer.fit_transform(combined_embeddings)

plt.figure(figsize=(8, 6))
plt.scatter(embedding_2d[:, 0], embedding_2d[:, 1], c=df['cluster'], cmap='Set2', s=80)
for i, txt in enumerate(df['cluster']):
    plt.annotate(txt, (embedding_2d[i, 0], embedding_2d[i, 1]))
plt.title("개별 문장 임베딩 기반 클러스터링 (UMAP 시각화)")
plt.show()

# 결과 확인
print(df[['profile', 'episode', 'cluster']])
```
